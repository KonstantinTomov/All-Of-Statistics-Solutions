**Question.** Suppose a coin has probability $p$ of falling heads up. If we flip the coin many times, we would expect the proportion of heads to be near $p$. We will make this formal later. Take $p = .3$ and $n = 1{,}000$ and simulate $n$ coin flips. Plot the proportion of heads as a function of $n$. Repeat for $p = .03$.

```python
import matplotlib.pyplot as plt
import random

Y = []
X = []
p = 0.3
n = 1000

numHeads = 0
for i in range(0,n,1):
    if random.random() < p:
        numHeads += 1
    Y.append(numHeads/(i+1))
    X.append(i)

plt.plot(X,Y,'r--')
plt.xlabel('Flips')
plt.ylabel('Proportion Heads')
plt.grid(True)
plt.show()

```
