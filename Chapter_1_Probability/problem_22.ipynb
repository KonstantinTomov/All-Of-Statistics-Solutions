**Question.** Suppose we flip a coin $n$ times and let $p$ denote the probability of heads. Let $X$ be the number of heads. We call $X$ a binomial random variable, which is discussed in the next chapter. Intuition suggests that $X$ will be close to $np$. To see if this is true, we can repeat this experiment many times and average the $X$ values. Carry out a simulation and compare the average of the $X$'s to $np$. Try this for $p = .3$ and $n = 10$, $n = 100$, and $n = 1{,}000$.


```python
import random
import statistics

p = 0.3
runs = 20
n = 10

results = []
for j in range(0,runs,1):
    X=0
    for i in range(0,n,1):
        if random.random() < p:
            X += 1
    results.append(X)

print(f"Over {runs} runs, each consisting of {n} flips, we got {results} number of heads in each run,")
print(f"this averages out to {statistics.mean(results)} (expected value is n*p={n*p})")
```
