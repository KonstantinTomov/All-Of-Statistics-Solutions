```python
import random
import statistics

p = 0.3
runs = 20
n = 10

results = []
for j in range(0,runs,1):
    X=0
    for i in range(0,n,1):
        if random.random() < p:
            X += 1
    results.append(X)

print(f"Over {runs} runs, each consisting of {n} flips, we got {results} number of heads in each run,")
print(f"this averages out to {statistics.mean(results)} (expected value is n*p={n*p})")
```
