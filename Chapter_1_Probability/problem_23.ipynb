```python
import random

die = [1,2,3,4,5,6]
A = [2,4,6]
B = [1,2,3,4]

n=1000
numA = 0
numB = 0
numAB = 0
for i in range(0,n,1):
    roll = random.choice(die)
    if roll in A:
        numA += 1
    if roll in B:
        numB += 1
    if roll in A and roll in B:
        numAB += 1

PA = numA/n
PB = numB/n
PAB = numAB/n

print(f"P(A) = {PA}, P(B)={PB}, P(AB)={PAB} and P(A)P(B)={PA*PB}")
```

We expected:

* $P(A) = \frac{1}{2} = 0.5$
* $P(B) = \frac{2}{3} \approx 0.667$
* $P(AB) = P(A)P(B) = \frac{1}{3} \approx 0.333$

The values we see from the simulation are very close to the expected ones
